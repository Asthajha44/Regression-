Q1. What is Simple Linear Regression?

Ans. Simple Linear Regression is a statistical metjod used to model and examine the relationship between two continuous variables:

* Dependent variables(X)- The outcome or variable you are trying to predict.
* Independent variable(Y)- The factor used to predict the dependent variable.

The goal of simple linear regression is to fit a straight line, known as the Line of Best Fit, through a set of data points to minimize the prediction error. The relationship between X and Y is linear. Observations are independent of one another.



Q2. What are the key assumptions of Simple Linear Regression?

Ans. Key assumptions of Simple Linear Regression:

1. Linearity - The relationship between the independent variable(X) and dependent variable(Y) must be linear. This means the change in Y for a unit change in X is constant, forming a straight line when plotted.
2. Independence - The observations must be independent of one another.One data point should not influence or depend on another.
1. Homoscedasticity - The spread of the residuals(the errors or distances between actual data points and the regression line) should be consistent across all values of X.The error should not get wider or narrower as X increases.
2. Normality of Errors - The error terms(residuals) should not follow a normal distribution. In simple terms, if you plot the errors, they should form a symmetrical bell-shaped curve centered around zero.





Q3. What does the coefficient m represent in the equation Y=mX+c?

Ans. In the equation y=mx+c, the coefficient m represents the slope of the line.

Coefficient m represents:-

* Rate of change- m measures how much the dependent variable(y) changes for every single-unit increase in the independent variable(x). Mathematically, it is calculated as change in y divided by change in x.
* Direction of line- If m is positive, the line slopes upward from left to right, meaning as x increases, y also increases. If m is negative, the line slopes downward from left to right, meaning as x increases, y decreases.
* Steepness of the line- The size of the number tells you how steep the line is. A larger number (e.g., m=5) means a steep, fast-rising line, while a smaller number close to zero (e.g., m=0.2) means a flat, slow-rising line. A slope of zero (m=0) means a completely horizontal line.
* Constant Growth/Decay- In a linear equation, m stays constant. This means the rate at which y changes realtive to x is uniform and never changes anywhere along the straight line.





Q4. What does the intercept c represent in the equation Y=mX+c?

Ans. In the equation y=mx+c, the intercept c represents the y intercept.

* Value when x=0: It is exact value of the dependent variable(y) when the independent variable(x) is zero.
* Baseline/Starting point: It serves as the initial baseline value before any changes or effects from x are applied.
* Graph Intersection: On a plotted grph, it is the exact point where the straight line crosses the vertical y-axis.
* Fixed Offset: It is constant value that shifts the entire line up or down on the graph without altering its slope or steepness.





Q5. How do we calculate the slope m in Simple Linear Regression?

Ans. In simple linear regression, the slope(m) is calculated using the Least Squares Method formula, which minimizes prediction errors.

Formula for slope(m): The slope (m) is equals to the sum of the products of each X minus its mean and each Y minus its mean, divided by the sum of the squared differences between each X and its mean.

Four key components of the calculation:

* Numerator (Covariance Component) measures how much X and Y change together relative to their means.
* Denominator (Variance Component) measures the total squared spread (variance) of the independent variable X around its mean.
* Ratio of Variation: dividing the combined variation of X and Y by the variation of X alone gives the exact rate of change.
* Minimizing Error: This specific calculation ensures the resulting line has the absolute minimum total distance (residuals) from all data points.








Q6. What is the purpose of the least squares method in Simple Linear Regression?

Ans. The primary purpose of the least squares method is to find the absolute best-fitting straight line through a set of data points by making the prediction errors as small as possible.

1. Finding the "Line of Best Fits"- It provides a mathematical way to draw a single staright line that comes closest to all the scattered data points on a graph.
2. Minimizing errors- It squares the vertical distances (errors) between the actual data points and the predicted line, and ensures the total sum of these squared errors is kept to an absolute minimum.
3. Calculating Exact Parameters- It provides the precise formulas needed to calculate exact values for the slope(m) and the intercept (c).
2. Making Accurate Predictions- By reducing the overall error, it ensures that the regression model makes future predictions as reliable and accurate as possible.





Q7. How is the coefficient of determination (R²) interpreted in Simple Linear Regression?

Ans. The coefficient of determination (R^2) measures how well the independent variable (X) explains the variaion in the dependent variable (Y).

* Percentage of Explained Variation- R^2 represents the percentage of total variation in Y that is explained by X.
(For ex- if R^2=0.85, it means 85% of the variation in Y is explained by X).
* Scale of 0 to 1 (or 0% to 100%)- It always ranges between O and 1. A value of 0 means the model explains nothing, while 1 (100%) means the model explains all the variability perfectly.
* Goodness of Fit- A highre R^2 value (closer to 1) indicates a strong "goodness of ift", meaning the regression line fits the data well and predictions will be reliable.
* Relation to Correlation- In simple linear regression, R^2 is simply the square of the correlation coefficient (r) between X and Y.





Q8. What is Multiple Linear Regression?

Ans. Multiple Linear Regression is an extension of simple linear regression that uses two or more independent variables (X1,X2,X3, etc.) to predict a single dependent variable (Y). It models the relationship between one continuous dependent variable and multiple independent predictor variables.

* Real World Practicality: Unlikr simple linear regression which only looks at one cause, multiple linear regression is much more realistic for the real world (e.g., predicting a house's price using square footage, number of bedrooms, and location all at once).
* Holding Constants (Partial effects): Each coefficient measures the change in Y for a one-unit change in that specific X, while holding all other variables constant.



Q9. What is the main difference between Simple and Multiple Linear Regression?

Ans. The core difference is the number of independent (predictor) variables used to predict the dependent variable.Simple linear regression uses only one independent variable to explain or predict a single dependent variable, examining a direct one-to-one relationship. In contrast, multiple linear regression uses two or more independent variables simultaneously to predict the dependent variable, making it much more ralistic for analyzing complex real-world situations where an outcome is influenced by multiple factors at the same time.



Q10. What are the key assumptions of Multiple Linear Regression?

Ans. Key assumptions of Multiple Linear Regression:-

1. Linearity- The relationship between each independent variable (X) and the dependent variable (Y) must be a straight-line relationship.
2. Independence- The observations and error terms must be independent of one another, meaning one data point should not predict or influence another.
3. Homoscedasticity (Constant Variation)- The spread of the residuals (prediction errors) must remain consistent and even across all values of th predictor variables, rather than widening or narrowing.
4. No Multicollinearity- The independent variables should not be too strongly correlated with each other. If predictors are closely tied together, it becomes difficult to figure out the individual impact of each variable on the outcome.



Q11. What is heteroscedasticity, and how does it affect the results of a Multiple Linear Regression model?

Ans. Heteroscedasticity happens when the variance (spread) of the error terms (residuals) is not constant across alll values of the independent variables. The prediction errors get wider and narrower as the values increase (often forming a megaphone shape on a scatter plot), violating the key assumption of homoscedasticity.

How it affects Multiple linear regression results:

* Coefficients Remain Unbiased- The calculated slope coefficients are still correct in average (unbiased), meaning the model's line is centered properly.
* Loss of Efficiency- The estimates are no longer the most precise or efficient ones possible. Other methods could give more accurate predictions.
* Distorted Standard Errors- The standard error of the coefficients are usually calculated incorrectly (often underestimated), which throws off the entire measure of reliability.
* Unreliable Hypothesis Tests (t-values and p-values)- Because standard errors are wrong, the t-statistics and p-values become misleading. This can lead you to falsely conclude that a variable is statistically significant (or insignificant) when it is actually the opposite.





Q12. How can you improve a Multiple Linear Regression model with high multicollinearity?

Ans. Here are four ways:-

* Drop Redundant Variables: Identify and remove one of the heavily correlated predictors usinga correlation matrix or Variance Inflation Factor (VIF).
* Combine Variables (Feature Engineering): Merge related variables into a single composite score or use dimensionality reduction techniques like PCA (Principle Component Analysis).
* Apply Regularization: Use Ridge or Lasso regression, which adds a penalty to coefficients to keeep the model stable despite correlated inputs.
* Collect More Data: Increase the sample size to lower sampling variance and help the model isolate unique effect of each variable.


Q13. What are some common techniques for transforming categorical variables for use in regression models?

Ans. Techniques for transforming categorical variables:-

1. One-Hot Encoding (Dummy Coding)- Splits a categorical variable into multiple binry columns (containing 0 or 1) representing the presence or absence of each category. (To avoid multicollinearity, one column is usually dropped).
2. Label Encoding- Assigns a unique sequential integer (e.g., 0,1,2,3) to each category. This is the best used for ordinal variables that have a clear, natural order (like low, medium or high).
3. Target (Mean) Encoding- Replaces each category with the average value of the dependent variable (Y) fpor that specific category, which is useful for categories with high cardinality (many unique groups).
4. Frequency (Count) Encoding- Replaces categories with the number of times (or percentage frequency) they appear in the dataset, ca[pturing how common or rare each category is.





Q14. What is the role of interaction terms in Multiple Linear Regression?

Ans. Role of Interaction Terms in Multiple Linear Regression:

An interaction term is used when the effect of one predictor variable on the dependent variable depends on the value or level of another predictor.

* Capturing combined effects- It measure the joint effect of two variables working together which cannot be seen by looking at each variable individually.
* Relaxing the additive assumption- Standard regression assumes variables act independently. Interaction terms allow the model to show that the impact of one factor changes depending on another.
* Improving model accurac- By accounting for complex real-world relationships, it reduces errors and providddes a much more accurate fit to the data than a standard model.
* Equation representation- It adds a new coefficient to the model to explicitly measures this moderating effect.





Q15. How can the interpretation of intercept differ between Simple and Multiple Linear Regression?

Ans. Difference in intercept interpretation: While the intercept acts as a baseline starting point in both models, its condition changes based on the number of variables:-

* Simple Linear Regression- It represents the expected value of Y when the isngle independent variable (X) is equal to zero.
* Multiple Linear Regression- It represents the expected value of Y when all independent variables (X1,X2,X3, etc.) are simultaneously equal to zero, holding all other variables constant.
* Practical Realism- In multiple regression, setting all predictors to zero at once is often physically or practically impossible, making the intercept in multiple regression primarily a mathematical anchor rather than a realistic scenario.







Q16. What is the significance of the slope in regression analysis, and how does it affect predictions?

Ans. The slope is a foundational component of a regression model. It measures the steepness and direction of the regression line.


1. Quantifies the Rate of Change: The slope tells you exact amount the dependent variable(Y) is expected to change for every single-unit increase in the independent variable(X).
2. Indicates the Direction of the Relationship: A positive slope means X and Y increase together, while a negative slope means as X increases, Y decreases.A slope of zero indicates no linear relationship.
3. Acts as the Core Multiplier in Predictions: In the prediction equation, the slope is multiplied by the given value of X, making it the primary mathematical driver for calculating expected or forecasted outcomes.
4. Determines Prediction Sensitivity (Steepness): A steep slope means that small changes in X will cause massive change in the predicted value of Y. Conversely, a flat slope means predictions for Y remain relatively stable regardless of changes in X.
5. Isolates Inidvidual Impact (in Multiple Regression): In multiple linear regression, the slope represents the unique effect of a specific predictor on the outcome while holding all other variables constant, ensuring precise and isolated predictions.



Q17. How does the intercept in a regression model provide context for the relationship between variables?

Ans. The intecept provides essential baseline context for understanding the relationship between variables in the following ways:-

* Establishes the baseline value- It tells you the expected value of the dependent variable(Y) when all independent variables (X) are zero, serving as a foundational starting point.
* Anchors the model graphically- It shifts the regression line vertically up or down on the coordinate plane, determining where the relationship officially beigins relative to the axes.
* Offers real-world meaning- In practical scenarios, it can represent a fixed initial cost, a base level, or a starting condition before any predictor variables have an effect (ex- base salary before experience is factored in).
* Serves as a reference point for the slope- By fixing  where line starts, it allows the slope coefficients to accurately measure changes away from this baseline rather than absolute magnitudes.



Q18. What are the limitations of using R² as a sole measure of model performance?

Ans. Limitations of using R^2 as a sole measure:-

* Increases with added variables (overfitting): R^2 never decreases when you add more independent variables-even if they are completely useless. This can trick you into thinking a complex model is better when it is actually just overfitted.
* Does not check model quality or assumptions: A high R^2 does not mean the model's assumptions are met, nor does it guarantee good out-of-sample predictions.
* Ignores residual patterns: It measures total variance explained but cannot tell you if there are hidden, systematic errors or no-linear patterns left in the residuals.





Q19. How would you interpret a large standard error for a regression coefficient?

Ans. A large standard error for a regression coefficient indicates a high degree of uncertainty and imprecision in estimating the true effect of that independent variable.

* High Imprecisions: It shows that the sample estimate of the coefficient varies widely and may not accurately reflect the true population relationship.
* Leads to Statistical Insignificance (t-test failure): Since the t-statistic is calculated as the coefficient divided by its standard error, a large denomiantor shrinks the t-value. This often results in a high p-value, causing you to incorrectly conclude that the variable is insignificant.
* Creates Wide Confidence Intervals: It results in a very wide confidence interval for the coefficient, meaning there is a broad range of plausible values for the variables's true impact.
* Signals Underlying Issues: A large standard error is often a symptom of problems in the data or model, such as multicollinearity, a small sample size, or high random noise.



Q20. How can heteroscedasticity be identified in residual plots, and why is it important to address it?

Ans. Identification via residual plots- On a scatter plot of residulas versus predicted values, heteroscedasticity is spotted by a distinct "cone" or "megaphone" shape, where the spread of the errors widens or narrows as values increase, rather than remaining evenly and randomly scattered.

It is important to fix because it causes standard errors to be calculated incorrectly, which creates a false sense of precision in the model.

It invalidates Hypothesis Test (t and p-values): Unchecked heteroscedasticity corrupts t-statistics and p-values, leading to misleading conclusions about whether an independent variable is truly significant.

Q21. What does it mean if a Multiple Linear Regression model has a high R² but low adjusted R²?

Ans. When a multiple linear regression model shows a high R^2 alongside a low adjusted R^2, it is classic warning sign of an overfitted model cluttered with useless variables.

* It indicates that many of the independent variables added to the model do not genuinely explain the dependent variable, they are just adding random noise.
* Regular R^2 automatically increases every time you add a new variable (even a useless one).In contrast, adjusted R^2 penalizes the model for adding variables that don't improve it significantly. A low adjusted R^2 exposes that the high regular R^2 is artificially inflated.
* Poor Generalization on New Data- While the model fits the current training data very well, its actual predictive power is weak, and it will likely perform poorly when applied to new, unseen data.

Q22. Why is it important to scale variables in Multiple Linear Regression?

Ans. It is important because of following reasons:-

* Enables Fair Comparison of Variable Importance- Independent variables often have vastly different units. Scaling them to a common range allows you to directly compare regression coefficients to see which variable has the strongest impact on the outcome.
* Essential for regularization (Ridge and Lasso)- Regularized regression models add a penalty to large coefficients. If predictors are unscaled, variables measured in smaller units will naturally have much larger coefficients and get penalized unfairly.
* Improves Optimization Efficiency (Gradient Descent)- If the model is solved using gradient descent, variables on different scales create skewed, uneven cost function surfaces, causing the algorithm to oscillate and take much longer to converge.





Q23. What is polynomial regression?

Ans. Polynomial regression is a specialized form of regression analysis used when the relationship between the independent and dependent variables is curved rather than a straight line. It captures these non-linear patterns by introducing higher-order powers of the predictor variables, such as squared or cubed terms, into the model. Despite producing a curved line of best fit, it is still categorized as a linear regression model because it remains linear in its unknown coefficients using standard least squares methods.

Q24. How does polynomial regression differ from linear regression?

Ans. Polynomial regression differs from standard linear regression primarily in the shape of the relationship it models. While standard linear regression assumes a constant, straight line relationship, polynomial regression introduces higher order powers of the predictor to capture complex, curved patterns. Consequently, its equation includes additional squared or cubed terms, allowing it to accurately model accelerating or decelerating rates of change that a simple straight line cannot represent.

Q25. When is polynomial regression used?

Ans. Polynomial regression is used in following scenarios:

* Non-Linear Relationship: When a scatter plot shows that the relationship between the independent (X) and dependent (Y) variables is curved (such as a U-shape or an inverted U-shape) rather than a straight line.
* Failure of Standard Linear Models: When a standard linear regression model underfits the data, resulting in systematic patterns remaining in the residuals because a straight line cannot capture the true underlying trend.



Q26. What is the general equation for polynomial regression?

Ans. General Equation for Polynomial Regression:
Dependent Variable = Intercept + (First Coefficient x Predictor) + (Second Coefficient x Predictor Squared) + (nth Coefficient x Predictor to the power of n) + Error Term

* The Outcome (Dependent Variable): The final predicted value or target outcome you are trying to forecast.
* The Baseline and Coefficients: The intercept serves as the starting baseline, while each coefficient measures the magnitude and direction of how that specific power of the predictor imapacts the curve.
* The Powers and Error Term: Raising the independent variable to higher powers allows the model to capture curved, non-linear relationships ending with the Error Term for random, unexplained variation.





Q27. Can polynomial regression be applied to multiple variables?

Ans. Yes, lynomial regression can be applied to multiple variables (often reffered to as multivariate polynomial regression).

* Higher-order terms of individual predictors to capture individual non-linear curves.
* Interaction terms to account for how multiple predictors jointly influence the dependent variable in a curved, multi-dimensional space.



Q28. What are the limitations of polynomial regression?

Ans. Limitations of Polynomial Regression:-

1. High Risk of Overfitting- Using a high polynomial degree allows the model to capture random noise in the training data rather than the true underlying trend, leading to wild oscillations and poor generalization on new data.
2. Exytremely Poor Extrapolation- Polynomial models perform terribly when making predictions outside the range of the original training data, often producingly wildly inaccurate, explosive forecasts as values move away from the observed dataset.
3. Severe Multicollinearity- Because highr-order terms are derived directly from the original predictor, they create high correlation among the variables, which drastically inflates standard errors and destablizes coefficient estimates.




Q29. What methods can be used to evaluate model fit when selecting the degree of a polynomial?

Ans. Methods to evaluate model fit fr polynomial degree selection:-

* Penalized Information Criteria (AIC and BIC): Metrics like the Akaike nformation Criterion (AIC) and Bayesian Information Criterion (BIC) measure the model fit while adding a strict penalty for higher degrees, helping identify the optimal balance between accuracy and complexity.
* k-Fold Cross Validation: By repeatedly partitioning the data into training and validation folds, this method measures true out-of-sample prediction error, allowing you to select the degree that yields the lowest generalization error.
* Validation Set Error (MSE/RMSE): Splitting data into a training set and a seperate validation set to track Mean squared Error(MSE), the ideal polynomial degree is found right before the validation error starts rising due to overfitting.



Q30. Why is visualization important in polynomial regression?

Ans. Visualization is important in Polynomial Regression because:-

* Detects Overfitting and Wild Oscillations: Plotting the fitted curve allows you to instantly see if a high-degree model is making erratic, unnatural swings between data points instead of capturing a smooth, true trend.
* Highlights Dangerous Extrapolation Limits: A graph clearly reveals the boundaries of the observed data, showing how quickly and wildly the polynomial curve explodes or behaves irrationally outside of that training range.
* Verifies Non-Linear Fit Quality: Overlaying the curve on a scatter plot helps you visually confirm whether your chosen degree actually matches the underlying shape (like a U-shape) or is merely chasing random noise.
